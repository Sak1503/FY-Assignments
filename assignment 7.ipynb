```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Accept CSV filename from the user
filename = input("Enter CSV filename: ")

try:
    # Load the dataset
    df = pd.read_csv(filename)

    # Validate required columns
    required_columns = [
        "Student", "Maths", "Python", "Attendance", "City"
    ]

    if not all(col in df.columns for col in required_columns):
        raise ValueError("CSV file is missing required columns.")

    if df.empty:
        raise ValueError("CSV file is empty.")

    # Display dataset
    print("\nDataset:")
    print(df)

    # Convert numerical columns to numeric values
    numeric_columns = ["Maths", "Python", "Attendance"]
    df[numeric_columns] = df[numeric_columns].apply(
        pd.to_numeric, errors="raise"
    )

    # 1. Line Plot
    plt.figure(figsize=(7, 4.5))
    plt.plot(df["Student"], df["Python"],
             marker="o", color="blue", label="Python")
    plt.title("Python Marks of Students")
    plt.xlabel("Student")
    plt.ylabel("Python Marks")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # 2. Bar Plot
    plt.figure(figsize=(7, 4.5))
    plt.bar(df["Student"], df["Maths"],
            color="orange", label="Maths")
    plt.title("Maths Marks of Students")
    plt.xlabel("Student")
    plt.ylabel("Maths Marks")
    plt.legend()
    plt.tight_layout()
    plt.show()

    # 3. Scatter Plot
    plt.figure(figsize=(7, 4.5))
    plt.scatter(df["Maths"], df["Python"],
                s=70, color="green", label="Students")
    plt.title("Maths vs Python Marks")
    plt.xlabel("Maths Marks")
    plt.ylabel("Python Marks")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # 4. Correlation Matrix and Heatmap
    correlation = df[numeric_columns].corr()

    print("\nCorrelation Matrix:")
    print(correlation)

    plt.figure(figsize=(6.5, 5))
    sns.heatmap(correlation, annot=True, cmap="Blues",
                vmin=-1, vmax=1, fmt=".2f")
    plt.title("Correlation Heatmap")
    plt.tight_layout()
    plt.show()

    # 5. Histogram of Python Marks
    plt.figure(figsize=(7, 4.5))
    plt.hist(df["Python"], bins=5, color="purple",
             edgecolor="black", label="Python Marks")
    plt.title("Distribution of Python Marks")
    plt.xlabel("Python Marks")
    plt.ylabel("Number of Students")
    plt.legend()
    plt.tight_layout()
    plt.show()

    # 6. Box Plot for Maths and Python
    plt.figure(figsize=(7, 4.5))
    sns.boxplot(data=df[["Maths", "Python"]])
    plt.title("Box Plot of Maths and Python Marks")
    plt.xlabel("Subject")
    plt.ylabel("Marks")
    plt.tight_layout()
    plt.show()

    # 7. Pie Chart of Students by City
    city_counts = df["City"].value_counts()

    plt.figure(figsize=(6, 6))
    plt.pie(city_counts, labels=city_counts.index,
            autopct="%1.1f%%", startangle=90)
    plt.title("Students from Each City")
    plt.tight_layout()
    plt.show()

    # 8. Calculate averages
    print("\nAverage Marks and Attendance:")
    print("Average Maths Marks:",
          round(df["Maths"].mean(), 2))
    print("Average Python Marks:",
          round(df["Python"].mean(), 2))
    print("Average Attendance:",
          round(df["Attendance"].mean(), 2))

except FileNotFoundError:
    print("Error: CSV file not found.")

except (ValueError, TypeError) as e:
    print("Error:", e)

except pd.errors.ParserError:
    print("Error: Invalid CSV file format.")
```
